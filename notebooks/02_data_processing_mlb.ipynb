```python
def clean_mlb_batting(df):
    
    # Keep the columns we need
    sum_cols = [
        "G", "AB", "R", "H", "2B", "3B", "HR",
        "RBI", "SB", "CS", "BB", "SO", "IBB",
        "HBP", "SH", "SF", "GIDP"
    ]
    
    # Combine multiple team stints into one player-season
    result = (
        df.groupby(["playerID", "yearID"], as_index=False)
        .agg({
            **{col: "sum" for col in sum_cols},
            "teamID": lambda x: "; ".join(x.astype(str).unique())
        })
    )
    
    # Calculate rate statistics
    result["BA"] = result["H"] / result["AB"]
    
    result["OBP"] = (
        result["H"] + result["BB"] + result["HBP"]
    ) / (
        result["AB"] + result["BB"] + result["HBP"] + result["SF"]
    )
    
    result["SLG"] = (
        result["H"]
        + result["2B"]
        + 2 * result["3B"]
        + 3 * result["HR"]
    ) / result["AB"]
    
    result["OPS"] = result["OBP"] + result["SLG"]
    
    return result
```

```python
mlb_batting_clean = clean_mlb_batting(mlb_batting)
```

```python
def clean_mlb_pitching(df):
    
    sum_cols = [
        "W", "L", "G", "GS", "CG", "SHO", "SV",
        "IPouts", "H", "ER", "HR", "BB", "SO",
        "IBB", "WP", "HBP", "BK", "BFP", "GF",
        "R", "SH", "SF", "GIDP"
    ]
    
    result = (
        df.groupby(["playerID", "yearID"], as_index=False)
        .agg({
            **{col: "sum" for col in sum_cols},
            "teamID": lambda x: "; ".join(x.astype(str).unique())
        })
    )
    
    # Convert outs to innings
    result["IP"] = result["IPouts"] / 3
    
    # Avoid division by zero for pitchers with no recorded outs
    ip = result["IP"].replace(0, np.nan)
    bb = result["BB"].replace(0, np.nan)
    decisions = (result["W"] + result["L"]).replace(0, np.nan)

    result["WPCT"] = result["W"] / decisions
    result["ERA"] = result["ER"] * 9 / ip
    result["WHIP"] = (result["BB"] + result["H"]) / ip
    result["K/9"] = result["SO"] * 9 / ip
    result["BB/9"] = result["BB"] * 9 / ip
    result["HR/9"] = result["HR"] * 9 / ip
    result["H/9"] = result["H"] * 9 / ip
    result["R/9"] = result["R"] * 9 / ip
    result["K/BB"] = result["SO"] / bb
    
    return result
```

```python
START_YEAR = 1995
END_YEAR = 2024

mlb_batting_clean = mlb_batting_clean[
    mlb_batting_clean["yearID"].between(START_YEAR, END_YEAR)
].copy()

mlb_pitching_clean = mlb_pitching_clean[
    mlb_pitching_clean["yearID"].between(START_YEAR, END_YEAR)
].copy()
```

```python
mlb_player_ids = pd.concat([
    mlb_batting_clean["playerID"],
    mlb_pitching_clean["playerID"]
]).unique()

mlb_people_clean = mlb_people[
    mlb_people["playerID"].isin(mlb_player_ids)
].copy()
```
