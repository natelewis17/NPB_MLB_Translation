```python
batting_sum_cols = [
    "G", "PA", "AB", "R", "RBI", "H", "1B", "2B", "3B",
    "HR", "TB", "SB", "CS", "BB", "IBB", "K", "DP", "DB",
    "SH", "SF"
]

batting_player_season = (
    batting_regular
    .groupby("PlayerID", as_index=False)
    .agg({
        "Season": "first",
        "Name": "first",
        "Position": "first",
        **{col: "sum" for col in batting_sum_cols}
    })
)
```

```python
batting_player_season["BA"] = (
    batting_player_season["H"] / batting_player_season["AB"]
)

batting_player_season["OBP"] = (
    (
        batting_player_season["H"]
        + batting_player_season["BB"]
        + batting_player_season["IBB"]
        + batting_player_season["DB"]
    )
    /
    (
        batting_player_season["AB"]
        + batting_player_season["BB"]
        + batting_player_season["IBB"]
        + batting_player_season["DB"]
        + batting_player_season["SF"]
    )
)

batting_player_season["SLG"] = (
    batting_player_season["TB"] / batting_player_season["AB"]
)

batting_player_season["OPS"] = (
    batting_player_season["OBP"] + batting_player_season["SLG"]
)
```

```python
def clean_npb_batting(df):
    df = df[df["Game Type"] == "Regular Season"].copy()
    sum_cols = [
        "G", "PA", "AB", "R", "RBI", "H", "1B", "2B", "3B",
        "HR", "TB", "SB", "CS", "BB", "IBB", "K", "DP", "DB",
        "SH", "SF"
    ]
    result = (
        df.groupby("PlayerID", as_index=False)
        .agg({
            "Season": "first",
            "Name": "first",
            "Position": "first",
            "Team": lambda x: "; ".join(x),
            **{col: "sum" for col in sum_cols}
        })
    )
    result["BA"] = result["H"] / result["AB"]
    result["SLG"] = result["TB"] / result["AB"]
    result["OBP"] = (
        result["H"] + result["BB"] + result["DB"]
    ) / (
        result["AB"] + result["BB"] + result["DB"] + result["SF"]
    )
    
    result["OPS"] = result["OBP"] + result["SLG"]
    
    return result
```

```python
batting_2015_clean = clean_npb_batting(batting_2015)
```

```python
def clean_npb_pitching(df):
    df = df[df["Game Type"] == "Regular Season"].copy()
    def ip_to_outs(ip):
        innings = int(ip)
        partial = round((ip - innings) * 10)
        return innings * 3 + partial

    df["Outs"] = df["IP"].apply(ip_to_outs)
    sum_cols = [
        "W", "L", "S", "HLD", "HP", "G",
        "BF", "H", "R", "ER", "HR", "K", "BB",
        "IBB", "DB", "WP", "BLK", "CG", "SHO",
        "No BB", "GS", "GF"
    ]
    result = (
        df.groupby("PlayerID", as_index=False)
        .agg({
            "Season": "first",
            "Name": "first",
            "Team": lambda x: "; ".join(x),
            **{col: "sum" for col in sum_cols},
            "Outs": "sum"
        })
    )
    result["IP"] = result["Outs"] / 3
    result["WPCT"] = (
        result["W"] /
        (result["W"] + result["L"])
    )

    result["ERA"] = (
        result["ER"] * 9 /
        result["IP"]
    )

    result["WHIP"] = (
        (result["BB"] + result["H"]) /
        result["IP"]
    )

    result["R/9"] = (
        result["R"] * 9 /
        result["IP"]
    )

    result["R/G"] = (
        result["R"] /
        result["G"]
    )

    result["H/9"] = (
        result["H"] * 9 /
        result["IP"]
    )

    result["HR/9"] = (
        result["HR"] * 9 /
        result["IP"]
    )

    result["K/9"] = (
        result["K"] * 9 /
        result["IP"]
    )

    result["BB/9"] = (
        result["BB"] * 9 /
        result["IP"]
    )

    result["K/BB"] = (
        result["K"] /
        result["BB"]
    )

    return result
```

```python
pitching_2015_clean = clean_npb_pitching(pitching_2015)

print("Rows:", len(pitching_2015_clean))
print("Unique players:", pitching_2015_clean["PlayerID"].nunique())
```

```python
pitching_2000_clean = clean_npb_pitching(pitching_2000)

print("Rows:", len(pitching_2000_clean))
print("Unique players:", pitching_2000_clean["PlayerID"].nunique())
```
