```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
PROJECT_ROOT = Path.cwd().parent
PROCESSED = PROJECT_ROOT / "data" / "processed"

hitters = pd.read_csv(PROCESSED / "npb_mlb_hitters.csv")
pitchers = pd.read_csv(PROCESSED / "npb_mlb_pitchers.csv")

print(hitters.shape)
print(pitchers.shape)
```

```python
hitters.columns.tolist()
```

```python
pitchers.columns.tolist()
```

```python
hitter_metrics = [
    "BA",
    "OBP",
    "SLG",
    "OPS"
]

hitter_comparison = pd.DataFrame({
    "NPB": [hitters[f"NPB_{metric}"].mean() for metric in hitter_metrics],
    "MLB": [hitters[f"MLB_{metric}"].mean() for metric in hitter_metrics]
}, index=hitter_metrics)

hitter_comparison
```

```python
pitcher_metrics = [
    "ERA",
    "WHIP",
    "K9",
    "BB9",
    "HR9"
]

pitcher_comparison = pd.DataFrame({
    "NPB": [pitchers[f"NPB_{metric}"].mean() for metric in pitcher_metrics],
    "MLB": [pitchers[f"MLB_{metric}"].mean() for metric in pitcher_metrics]
}, index=pitcher_metrics)

pitcher_comparison
```
